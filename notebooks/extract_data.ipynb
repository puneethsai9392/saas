```python
import os

# Make sure the Kaggle config directory exists
os.environ['KAGGLE_CONFIG_DIR'] = os.path.expanduser("~/.kaggle")

def download_kaggle_dataset(dataset_link: str, download_path: str = "./data"):
    """
    Download a Kaggle dataset given its URL.
    Example dataset_link: 'https://www.kaggle.com/datasets/iamsouravbanerjee/world-population-dataset'
    """
    # Extract dataset slug from link
    dataset_slug = dataset_link.split("/datasets/")[-1]
    
    # Ensure download directory exists
    os.makedirs(download_path, exist_ok=True)
    
    # Download the dataset using kaggle API
    os.system(f"kaggle datasets download -d {dataset_slug} -p {download_path} --unzip")
    
    print(f"✅ Download complete! Files are in: {download_path}")
```

```python
dataset_links = [
    'https://www.kaggle.com/datasets/ashikadnan/railway-track-fault-detection-dataset2fastener',
    'https://www.kaggle.com/datasets/salmaneunus/railway-track-fault-detection',
    'https://www.kaggle.com/datasets/ashikadnan/railway-track-fault-detection-dataset1-rail'
    ]

for link in dataset_links:
    download_kaggle_dataset(link)
```
